# Week 3 — Financial Data Visualization & Reporting

This notebook consumes the Week 2 cleaned quarterly panel for TCS, Infosys, HCLTech and Wipro.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
ROOT = Path('..').resolve()
DATA = ROOT / 'data' / 'cleaned' / 'quarterly_cleaned.csv'
OUT = ROOT / 'reports' / 'week3' / 'figures'
QUARTERS = ['Q1 FY24','Q2 FY24','Q3 FY24','Q4 FY24','Q1 FY25','Q2 FY25','Q3 FY25','Q4 FY25','Q1 FY26','Q2 FY26','Q3 FY26','Q4 FY26','Q1 FY27']
COMPANIES = ['TCS','Infosys','HCLTech','Wipro']
df = pd.read_csv(DATA)
df['Fiscal Quarter'] = pd.Categorical(df['Fiscal Quarter'], categories=QUARTERS, ordered=True)
df = df.sort_values(['Company','Fiscal Quarter'])
df.head()
```

## 1. Revenue trend

```python
plt.figure(figsize=(10.8,5.8))
ax = plt.gca()
for company in COMPANIES:
    s = df[df.Company == company].sort_values('Fiscal Quarter')
    ax.plot(range(len(QUARTERS)), s['Revenue (INR Cr)'], marker='o', linewidth=2, label=company)
ax.set_title('Quarterly Revenue Trend — Common Panel', fontsize=16, pad=14, fontweight='bold')
ax.set_ylabel('Revenue (₹ crore)')
ax.set_xticks(range(len(QUARTERS))); ax.set_xticklabels(QUARTERS, rotation=35, ha='right')
ax.grid(axis='y', alpha=.22); ax.legend(ncol=4, loc='upper left')
plt.tight_layout(); plt.show()
```

## 2. Indexed revenue and peer growth

```python
idx = df.copy(); idx['Revenue Index'] = idx['Revenue (INR Cr)'] / idx.groupby('Company')['Revenue (INR Cr)'].transform('first') * 100
plt.figure(figsize=(10.8,5.8)); ax=plt.gca()
for company in COMPANIES:
    s=idx[idx.Company==company].sort_values('Fiscal Quarter'); ax.plot(range(len(QUARTERS)), s['Revenue Index'], marker='o', linewidth=2, label=company)
ax.set_title('Revenue Index — Q1 FY24 = 100', fontsize=16, pad=14, fontweight='bold'); ax.set_ylabel('Index'); ax.set_xticks(range(len(QUARTERS))); ax.set_xticklabels(QUARTERS, rotation=35, ha='right'); ax.grid(axis='y', alpha=.22); ax.legend(ncol=4, loc='upper left'); plt.tight_layout(); plt.show()
```

## 3. Margin trends

```python
for metric, title, ylabel in [('Operating Margin (%)','Operating Margin Trend — Standardized Week 2 Panel','Panel Operating Margin (%)'),('Net Margin (%)','Net Margin Trend — Standardized Week 2 Panel','Net Margin (%)')]:
    plt.figure(figsize=(10.8,5.8)); ax=plt.gca()
    for company in COMPANIES:
        s=df[df.Company==company].sort_values('Fiscal Quarter'); ax.plot(range(len(QUARTERS)), s[metric], marker='o', linewidth=2, label=company)
    ax.set_title(title, fontsize=16, pad=14, fontweight='bold'); ax.set_ylabel(ylabel); ax.set_xticks(range(len(QUARTERS))); ax.set_xticklabels(QUARTERS, rotation=35, ha='right'); ax.grid(axis='y', alpha=.22); ax.legend(ncol=4, loc='lower left'); plt.tight_layout(); plt.show()
```

## Interpretation discipline

The historical panel uses standardized Week 2 fields. Wipro's issuer-reported Q1 FY27 IT-services operating margin and other current issuer-defined margins are not treated as interchangeable with the standardized historical panel margin. This distinction is carried into the Week 3 report.
